```python
def base_stat_info(array):
    import numpy as np
    lower_outliers = list()
    upper_outliers = []
    q1 = np.quantile(array, .25, method = 'weibull')
    q2 = np.quantile(array, .5, method = 'weibull')
    q3 = np.quantile(array, .75, method = 'weibull')
    iqr = q3 - q1
    out_lower_bound = q1 - 1.5 * iqr
    out_upper_bound = q3 + 1.5 * iqr
    print("Q1 = ", q1)
    print("Q2 = M = ", q2)
    print("Q3 = ", q3)
    print("Interquartile Range = ", iqr)
    print("Lower bound: ", out_lower_bound)
    print("Upper bound: ", out_upper_bound)
    for i in array:
        if i < out_lower_bound:
            lower_outliers.append(i)
        elif i > out_upper_bound:
            upper_outliers.append(i)
    print("Outliers: ", lower_outliers + upper_outliers)
    
    
```

```python
result = base_stat_info(df["Sleep_Duration"])
print(result)
```
